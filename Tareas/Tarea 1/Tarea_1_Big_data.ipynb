```python
import time
import psutil
import multiprocessing
from collections import defaultdict
import gzip
import os

file_path = './ecoli-k12-ref.fna'

def read_fasta_by_block(file_path, block_size=1000):
    open_func = gzip.open if file_path.endswith('.gz') else open
    with open_func(file_path, 'rt') as file:
        seq = []
        block = []
        for line in file:
            line = line.strip()
            if line.startswith('>'):
                if seq:
                    block.append(''.join(seq))
                    seq = []
                if len(block) >= block_size:
                    yield block
                    block = []
            else:
                seq.append(line)
        if seq:
            block.append(''.join(seq))
        if block:
            yield block

def count_kmers(sequence, k):
    kmer_counts = defaultdict(int)
    for i in range(len(sequence) - k + 1):
        kmer = sequence[i:i+k]
        kmer_counts[kmer] += 1
    return kmer_counts

def process_block(block, k_values):
    kmer_counts = {k: defaultdict(int) for k in k_values}
    for seq in block:
        for k in k_values:
            if len(seq) >= k:
                partial_counts = count_kmers(seq, k)
                for kmer, count in partial_counts.items():
                    kmer_counts[k][kmer] += count
    return kmer_counts

def store_partial_counts(kmer_counts, k, block_num):
    with open(f'partial_kmer_counts_k{k}_block{block_num}.txt', 'w') as f:
        for kmer, count in kmer_counts[k].items():
            f.write(f"{kmer}\t{count}\n")

def merge_counts(k_values, num_blocks):
    final_counts = {k: defaultdict(int) for k in k_values}

    for k in k_values:
        for block_num in range(num_blocks):
            partial_file = f'partial_kmer_counts_k{k}_block{block_num}.txt'
            with open(partial_file, 'r') as f:
                for line in f:
                    kmer, count = line.strip().split('\t')
                    final_counts[k][kmer] += int(count)
            os.remove(partial_file)

    return final_counts

def count_all_kmers(file_path, k_values=[10, 12], block_size=1000):
    block_num = 0
    num_blocks = 0

    for block in read_fasta_by_block(file_path, block_size):
        kmer_counts = process_block(block, k_values)

        for k in k_values:
            store_partial_counts(kmer_counts, k, block_num)

        block_num += 1
        num_blocks += 1

    return merge_counts(k_values, num_blocks)

def generate_frequency_report(kmer_counts, k, filename):
    frequency_distribution = defaultdict(int)

    counts = kmer_counts[k]
    for count in counts.values():
        if count > 100:
            count = 100
        frequency_distribution[count] += 1

    sorted_distribution = sorted(frequency_distribution.items())

    with open(filename, 'w') as file:
        for freq, num_kmers in sorted_distribution:
            file.write(f"{freq} {num_kmers}\n")

def main(file_path):
    start_time = time.time()

    k_values = [10, 12]
    block_size = 500
    kmer_counts = count_all_kmers(file_path, k_values, block_size)

    # genera informes separados para k=10 y k=12
    for k in k_values:
        report_filename = f'kmer_frequency_report_k{k}.txt'
        generate_frequency_report(kmer_counts, k, report_filename)
        print(f"Informe generado en: {report_filename}")

    end_time = time.time()


    execution_time = end_time - start_time
    process = psutil.Process()
    memory_usage = process.memory_info().rss / (1024 * 1024)
    num_cores = multiprocessing.cpu_count()

    print(f"Tiempo de ejecución: {execution_time:.2f} segundos")
    print(f"Uso de memoria: {memory_usage:.2f} MB")
    print(f"Número de núcleos: {num_cores}")

# ejecutar la funcion principal
main(file_path)
```

c) Implemente una versión secuencial que lea archivos fasta y cuente los 10-mers y 12-mers (opcionalmente 30-mers) usando C.

```python
%%writefile contador-kmers.c
#include <stdio.h>
#include <stdlib.h>
#include <string.h>

#define MAX_SEQ_LENGTH 10000

typedef struct {
    char kmer[13];  // asumimos que el k-mer mas largo tiene longitud 12
    int count;
} KmerCount;

KmerCount* kmers = NULL;
int kmer_count = 0;
int kmer_capacity = 0;

void add_kmer(const char* kmer) {
    for (int i = 0; i < kmer_count; i++) {
        if (strcmp(kmers[i].kmer, kmer) == 0) {
            kmers[i].count++;
            return;
        }
    }

    if (kmer_count >= kmer_capacity) {
        kmer_capacity = kmer_capacity ? kmer_capacity * 2 : 256;
        kmers = realloc(kmers, kmer_capacity * sizeof(KmerCount));
    }

    strcpy(kmers[kmer_count].kmer, kmer);
    kmers[kmer_count].count = 1;
    kmer_count++;
}

void count_kmers(const char* sequence, int k) {
    int len = strlen(sequence);
    for (int i = 0; i <= len - k; i++) {
        char kmer[13];  // Máximo tamaño de kmer es 12 + 1 para terminador null
        strncpy(kmer, &sequence[i], k);
        kmer[k] = '\0';
        add_kmer(kmer);
    }
}

int main(int argc, char* argv[]) {
    if (argc != 2) {
        fprintf(stderr, "Uso: %s archivo.fasta\n", argv[0]);
        return 1;
    }

    FILE* file = fopen(argv[1], "r");
    if (!file) {
        perror("Error abriendo archivo");
        return 1;
    }

    char line[MAX_SEQ_LENGTH];
    char sequence[MAX_SEQ_LENGTH] = "";

    // leer el archivo linea por linea
    while (fgets(line, sizeof(line), file)) {
        if (line[0] == '>') {
            continue;
        }
        line[strcspn(line, "\n")] = '\0';
        strcat(sequence, line);
    }

    fclose(file);

    // Contar 10-mers y 12-mers
    count_kmers(sequence, 10);
    count_kmers(sequence, 12);

    // Imprimir resultados
    for (int i = 0; i < kmer_count; i++) {
        printf("%s %d\n", kmers[i].kmer, kmers[i].count);
    }

    free(kmers);
    return 0;
}
```

d) Implemente una versión utilizando hilos que paralelice el conteo de k-mers mediante la división de tareas por secuencias, considere que cada hilo dispondrá de un buffer de X secuencias (por ejemplo, x=1000).

```python
%%writefile d.c
#include <stdio.h>
#include <stdlib.h>
#include <string.h>
#include <pthread.h>
#include <unistd.h>
#include <sys/resource.h>

#define BLOCK_SIZE 1000
#define MAX_K 12

typedef struct {
    char **sequences;
    int count;
} Block;

typedef struct {
    char *sequence;
    int k;
} KmerTask;

typedef struct {
    int k;
    pthread_mutex_t mutex;
    int kmer_counts[1000000];
} KmerCounter;

Block* read_fasta_by_block(const char *file_path, int block_size, int *num_blocks) {
    FILE *file = fopen(file_path, "r");
    if (!file) {
        perror("Error opening file");
        return NULL;
    }

    Block *blocks = malloc(sizeof(Block) * block_size);
    int block_count = 0;
    int seq_count = 0;
    char line[1024];
    Block *current_block = NULL;

    while (fgets(line, sizeof(line), file)) {
        line[strcspn(line, "\n")] = '\0'; // Remove newline character

        if (line[0] == '>') {
            if (seq_count > 0) {
                current_block->sequences[seq_count] = NULL;
                seq_count = 0;
                block_count++;
                if (block_count >= block_size) {
                    blocks = realloc(blocks, sizeof(Block) * (block_size * 2));
                    block_size *= 2;
                }
                current_block = &blocks[block_count];
                current_block->sequences = malloc(sizeof(char*) * block_size);
            }
        } else {
            if (seq_count >= block_size) {
                current_block->sequences = realloc(current_block->sequences, sizeof(char*) * (block_size * 2));
                block_size *= 2;
            }
            current_block->sequences[seq_count] = strdup(line);
            seq_count++;
        }
    }

    if (seq_count > 0) {
        current_block->sequences[seq_count] = NULL;
        block_count++;
    }

    fclose(file);
    *num_blocks = block_count;
    return blocks;
}

void* count_kmers(void *arg) {
    KmerTask *task = (KmerTask*) arg;
    char *sequence = task->sequence;
    int k = task->k;

    int kmer_counts[1000000] = {0};

    int len = strlen(sequence);
    for (int i = 0; i <= len - k; i++) {
        char kmer[k + 1];
        strncpy(kmer, sequence + i, k);
        kmer[k] = '\0';
    }

    free(task->sequence);
    free(task);
    pthread_exit(NULL);
}

void* process_block(void *arg) {
    Block *block = (Block*) arg;

    pthread_t threads[block->count];
    for (int i = 0; i < block->count; i++) {
        KmerTask *task = malloc(sizeof(KmerTask));
        task->sequence = strdup(block->sequences[i]);
        task->k = 10; // Example k-mer size, adjust as needed
        pthread_create(&threads[i], NULL, count_kmers, task);
    }

    for (int i = 0; i < block->count; i++) {
        pthread_join(threads[i], NULL);
    }

    for (int i = 0; i < block->count; i++) {
        free(block->sequences[i]);
    }
    free(block->sequences);
    free(block);

    pthread_exit(NULL);
}

void generate_frequency_report(const char *filename, int k) {
    FILE *file = fopen(filename, "w");
    if (!file) {
        perror("Error opening file");
        return;
    }
    int frequency_distribution[100] = {0};

    for (int i = 0; i < 100; i++) {
        fprintf(file, "%d %d\n", i, frequency_distribution[i]);
    }

    fclose(file);
}

int main(int argc, char *argv[]) {
    if (argc < 2) {
        fprintf(stderr, "Usage: %s <file_path>\n", argv[0]);
        return 1;
    }

    const char *file_path = argv[1];
    int num_blocks = 0;
    Block *blocks = read_fasta_by_block(file_path, BLOCK_SIZE, &num_blocks);

    pthread_t block_threads[num_blocks];
    for (int i = 0; i < num_blocks; i++) {
        pthread_create(&block_threads[i], NULL, process_block, &blocks[i]);
    }

    for (int i = 0; i < num_blocks; i++) {
        pthread_join(block_threads[i], NULL);
    }

    for (int k = 10; k <= 12; k += 2) {
        char filename[100];
        snprintf(filename, sizeof(filename), "d_frecuencia %d-mers.txt", k);
        generate_frequency_report(filename, k);
        printf("Informe generado en: %s\n", filename);
    }

    struct rusage usage;
    getrusage(RUSAGE_SELF, &usage);
    long rss = usage.ru_maxrss;

    printf("Uso de memoria: %ld KB\n", rss);
    printf("Número de núcleos: %ld\n", sysconf(_SC_NPROCESSORS_ONLN));

    free(blocks);
    return 0;
}


```

```python
%%writefile paralero.c
#include <stdio.h>
#include <stdlib.h>
#include <string.h>
#include <pthread.h>
#include <time.h>
#include <sys/sysinfo.h>

#define KMER_LENGTH_10 10
#define KMER_LENGTH_12 12
#define BUFFER_SIZE 1048576  // 1 MB
#define NUM_THREADS 4
#define MAX_FREQUENCY 100

typedef struct {
    char *sequence;
    int start;
    int end;
    int thread_id;
    int *kmer_frequencies_10;
    int *kmer_frequencies_12;
} ThreadData;

void count_kmers(const char *sequence, int kmer_length, int *kmer_frequencies) {
    int length = strlen(sequence);
    if (length < kmer_length) return;

    for (int i = 0; i <= length - kmer_length; ++i) {
        // Aquí se debe calcular un hash o índice para cada k-mer
        // Esto es solo un ejemplo básico, debes adaptar el cálculo del índice del k-mer
        unsigned long hash = 0;
        for (int j = 0; j < kmer_length; ++j) {
            hash = hash * 31 + sequence[i + j]; // Un simple hash basado en el k-mer
        }
        kmer_frequencies[hash % (length - kmer_length + 1)]++;
    }
}

void *thread_function(void *arg) {
    ThreadData *data = (ThreadData *)arg;
    char *seq = strndup(data->sequence + data->start, data->end - data->start);

    // Procesar los k-mers de longitud 10 y 12
    count_kmers(seq, KMER_LENGTH_10, data->kmer_frequencies_10);
    count_kmers(seq, KMER_LENGTH_12, data->kmer_frequencies_12);

    free(seq);
    return NULL;
}

void report_frequencies(int *frequencies, int kmer_length, int sequence_length, const char *filename) {
    FILE *output_file = fopen(filename, "w");
    if (!output_file) {
        printf("Error: No se pudo crear el archivo %s\n", filename);
        perror("fopen");
        return;
    }

    int freq_count[MAX_FREQUENCY + 1] = {0};

    // Contar las frecuencias de k-mers y limitar a 100
    for (int i = 0; i < sequence_length - kmer_length + 1; ++i) {
        int freq = frequencies[i];
        if (freq > MAX_FREQUENCY) freq = MAX_FREQUENCY;
        freq_count[freq]++;
    }

    // Escribir los resultados en el archivo
    for (int i = 1; i <= MAX_FREQUENCY; ++i) {
        if (freq_count[i] > 0) {
            fprintf(output_file, "%d %d\n", i, freq_count[i]);
        }
    }

    fclose(output_file);
}

int main(int argc, char *argv[]) {
    if (argc != 2) {
        fprintf(stderr, "Uso: %s <archivo_fasta>\n", argv[0]);
        return EXIT_FAILURE;
    }

    // Medir tiempo de inicio
    clock_t start, end;
    double cpu_time_used;
    start = clock();

    // Obtener número de núcleos disponibles
    int num_cores = get_nprocs();
    printf("Número de núcleos: %d\n", num_cores);

    // Leer el archivo FASTA
    FILE *file = fopen(argv[1], "r");
    if (file == NULL) {
        perror("No se pudo abrir el archivo");
        return EXIT_FAILURE;
    }

    char buffer[BUFFER_SIZE];
    char sequence[BUFFER_SIZE] = {0};
    while (fgets(buffer, sizeof(buffer), file)) {
        if (buffer[0] == '>') {
            continue;  // Ignorar las cabeceras FASTA
        }
        buffer[strcspn(buffer, "\r\n")] = 0;  // Eliminar saltos de línea
        strncat(sequence, buffer, sizeof(sequence) - strlen(sequence) - 1);
    }

    fclose(file);

    // Verificar la longitud de la secuencia
    if (strlen(sequence) == 0) {
        printf("Error: Secuencia vacía o no válida\n");
        return EXIT_FAILURE;
    }

    printf("Secuencia total leída: %lu caracteres\n", strlen(sequence));

    // Crear matrices para almacenar las frecuencias de k-mers
    int *kmer_frequencies_10 = (int *)calloc(strlen(sequence) - KMER_LENGTH_10 + 1, sizeof(int));
    int *kmer_frequencies_12 = (int *)calloc(strlen(sequence) - KMER_LENGTH_12 + 1, sizeof(int));

    // Crear los hilos
    pthread_t threads[NUM_THREADS];
    ThreadData thread_data[NUM_THREADS];
    int length = strlen(sequence);
    int chunk_size = length / NUM_THREADS;

    for (int i = 0; i < NUM_THREADS; ++i) {
        thread_data[i].sequence = sequence;
        thread_data[i].start = i * chunk_size;
        thread_data[i].end = (i == NUM_THREADS - 1) ? length : (i + 1) * chunk_size;
        thread_data[i].thread_id = i + 1;
        thread_data[i].kmer_frequencies_10 = kmer_frequencies_10;
        thread_data[i].kmer_frequencies_12 = kmer_frequencies_12;
        pthread_create(&threads[i], NULL, thread_function, &thread_data[i]);
    }

    for (int i = 0; i < NUM_THREADS; ++i) {
        pthread_join(threads[i], NULL);
    }

    // Generar reportes
    report_frequencies(kmer_frequencies_10, KMER_LENGTH_10, length, "paralero_frecuencia_10.txt");
    report_frequencies(kmer_frequencies_12, KMER_LENGTH_12, length, "paralero_frecuencia_12.txt");

    // Liberar memoria
    free(kmer_frequencies_10);
    free(kmer_frequencies_12);

    // Medir tiempo de finalización
    end = clock();
    cpu_time_used = ((double) (end - start)) / CLOCKS_PER_SEC;
    printf("Tiempo de ejecución: %f segundos\n", cpu_time_used);

    return EXIT_SUCCESS;
}

```

```python
%%writefile f3.c
#include <stdio.h>
#include <stdlib.h>
#include <string.h>
#include <pthread.h>
#include <time.h>
#include <sys/sysinfo.h>

#define KMER_LENGTH_10 10
#define KMER_LENGTH_12 12
#define BUFFER_SIZE 1048576  // 1 MB
#define NUM_THREADS 4
#define MAX_FREQUENCY 100

typedef struct {
    char *sequence;
    int start;
    int end;
    int thread_id;
    int *kmer_frequencies_10;
    int *kmer_frequencies_12;
} ThreadData;

void count_kmers(const char *sequence, int kmer_length, int *kmer_frequencies) {
    int length = strlen(sequence);
    if (length < kmer_length) return;

    for (int i = 0; i <= length - kmer_length; ++i) {
        unsigned long hash = 0;
        for (int j = 0; j < kmer_length; ++j) {
            hash = hash * 31 + sequence[i + j];
        }
        kmer_frequencies[hash % (length - kmer_length + 1)]++;
    }
}

void *thread_function(void *arg) {
    ThreadData *data = (ThreadData *)arg;
    char *seq = strndup(data->sequence + data->start, data->end - data->start);

    count_kmers(seq, KMER_LENGTH_10, data->kmer_frequencies_10);
    count_kmers(seq, KMER_LENGTH_12, data->kmer_frequencies_12);

    free(seq);
    return NULL;
}

void report_frequencies(int *frequencies, int kmer_length, int sequence_length, const char *filename) {
    FILE *output_file = fopen(filename, "w");
    if (!output_file) {
        printf("Error: No se pudo crear el archivo %s\n", filename);
        perror("fopen");
        return;
    }

    int freq_count[MAX_FREQUENCY + 1] = {0};

    for (int i = 0; i < sequence_length - kmer_length + 1; ++i) {
        int freq = frequencies[i];
        if (freq > MAX_FREQUENCY) freq = MAX_FREQUENCY;
        freq_count[freq]++;
    }

    for (int i = 1; i <= MAX_FREQUENCY; ++i) {
        if (freq_count[i] > 0) {
            fprintf(output_file, "%d %d\n", i, freq_count[i]);
        }
    }

    fclose(output_file);
}

void generate_filename(int kmer_length, const char* prefix, char* output_filename) {
    sprintf(output_filename, "%s_f3_%d.txt", prefix, kmer_length);
}

int main(int argc, char *argv[]) {
    if (argc != 2) {
        fprintf(stderr, "Uso: %s <archivo_fasta>\n", argv[0]);
        return EXIT_FAILURE;
    }

    clock_t start, end;
    double cpu_time_used;
    start = clock();

    int num_cores = get_nprocs();
    printf("Número de núcleos: %d\n", num_cores);

    FILE *file = fopen(argv[1], "r");
    if (file == NULL) {
        perror("No se pudo abrir el archivo");
        return EXIT_FAILURE;
    }

    char buffer[BUFFER_SIZE];
    char sequence[BUFFER_SIZE] = {0};
    while (fgets(buffer, sizeof(buffer), file)) {
        if (buffer[0] == '>') {
            continue;
        }
        buffer[strcspn(buffer, "\r\n")] = 0;
        strncat(sequence, buffer, sizeof(sequence) - strlen(sequence) - 1);
    }

    fclose(file);

    if (strlen(sequence) == 0) {
        printf("Error: Secuencia vacía o no válida\n");
        return EXIT_FAILURE;
    }

    printf("Secuencia total leída: %lu caracteres\n", strlen(sequence));

    int *kmer_frequencies_10 = (int *)calloc(strlen(sequence) - KMER_LENGTH_10 + 1, sizeof(int));
    int *kmer_frequencies_12 = (int *)calloc(strlen(sequence) - KMER_LENGTH_12 + 1, sizeof(int));

    pthread_t threads[NUM_THREADS];
    ThreadData thread_data[NUM_THREADS];
    int length = strlen(sequence);
    int chunk_size = length / NUM_THREADS;

    for (int i = 0; i < NUM_THREADS; ++i) {
        thread_data[i].sequence = sequence;
        thread_data[i].start = i * chunk_size;
        thread_data[i].end = (i == NUM_THREADS - 1) ? length : (i + 1) * chunk_size;
        thread_data[i].thread_id = i + 1;
        thread_data[i].kmer_frequencies_10 = kmer_frequencies_10;
        thread_data[i].kmer_frequencies_12 = kmer_frequencies_12;
        pthread_create(&threads[i], NULL, thread_function, &thread_data[i]);
    }

    for (int i = 0; i < NUM_THREADS; ++i) {
        pthread_join(threads[i], NULL);
    }

    char filename_10[50];
    char filename_12[50];

    generate_filename(KMER_LENGTH_10, "report_frequencies", filename_10);
    generate_filename(KMER_LENGTH_12, "report_frequencies", filename_12);

    report_frequencies(kmer_frequencies_10, KMER_LENGTH_10, length, filename_10);
    report_frequencies(kmer_frequencies_12, KMER_LENGTH_12, length, filename_12);

    free(kmer_frequencies_10);
    free(kmer_frequencies_12);

    end = clock();
    cpu_time_used = ((double)(end - start)) / CLOCKS_PER_SEC;
    printf("Tiempo de ejecución: %f segundos\n", cpu_time_used);

    return EXIT_SUCCESS;
}
```
